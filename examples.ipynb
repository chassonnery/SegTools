# Examples of segmentation with the SegViz tool

The script below contains examples of use of the segmentation part of the SegViz tool. They make it possible to reproduce all the figures of the SegViz user manual.

### Import statement

Execute the cell below first of all to make the function WatershedSegmentation available. The other cells can then be executed in any order.

```python
from SegTool import WatershedSegmentation
```

### Segmentation using the default settings

This enables to reproduce Figures 12 and 13 of the user manual.

```python
WatershedSegmentation("image", "demo_data/bananas.tiff", savefile="bananas_default") # Fig 12

WatershedSegmentation("dataset", "demo_data/default.csv") # Fig 13
```

### Segmentation using rod-like separators

This enables to reproduce Figures 14 and 15 of the user manual.

```python
WatershedSegmentation("image", "demo_data/bananas.tiff",
                      savefile="bananas_withsep",
                      InputRods="demo_data/bananas_separators.tiff") # Fig 14

WatershedSegmentation("dataset", "demo_data/default.csv",
                      savefile="default_seg_withsep",
                      InputRods="demo_data/default_separators.csv") # Fig 15
```

### Segmentation using different values of the "resolution" parameter

This enables to reproduce Figure 16 of the user manual.

```python
WatershedSegmentation("dataset", "demo_data/resolution.csv", savefile="resolution1", resolution=1)

WatershedSegmentation("dataset", "demo_data/resolution.csv", savefile="resolution5", resolution=5)
```

### Segmentation using different values of the domain-size parameters "xmax" and "ymax"

This enables to reproduce Figure 18 of the user manual.

The first two instructions should yield the exact same segmentation (i.e. the content of files domain_size_auto.csv and domain_size_auto2.csv should be identical), but the second will produce warnings because the parameters xmax, ymax and zmax are left to default.

The third and fourth instructions will also produce warnings for the same reason (parameter zmax left to default in both case).

The fifth case should yield the exact same segmentation than the fourth but will not produce any warning.

```python
WatershedSegmentation("dataset", "demo_data/domain_size.csv", savefile="domain_size_auto")

WatershedSegmentation("dataset", "demo_data/domain_size.csv",
                      savefile="domain_size_auto2",
                      PeriodicBoundaries=True)

WatershedSegmentation("dataset", "demo_data/domain_size.csv",
                      savefile="domain_size_xmax4.2",
                      PeriodicBoundaries=True, xmax=4.2) # This will print a warning 

WatershedSegmentation("dataset", "demo_data/domain_size.csv",
                      savefile="domain_size_xmax4.2_ymax3",
                      PeriodicBoundaries=True, xmax=4.2, ymax=3)

WatershedSegmentation("dataset", "demo_data/domain_size.csv",
                      savefile="domain_size_xmax4.2_ymax3_zmax3.5",
                      PeriodicBoundaries=True, xmax=4.2, ymax=3, zmax=3.5)
```

### Segmentation using different values of the dilatation parameter "dil_coeff"

This enables to reproduce Figure 19 of the user manual.

```python
WatershedSegmentation("dataset", "demo_data/scaling_up.csv",
                      savefile="scaling_up1") # default value of dil_coeff is 1.0

WatershedSegmentation("dataset", "demo_data/scaling_up.csv",
                      savefile="scaling_up1.2",
                      dil_coeff=1.2)

WatershedSegmentation("dataset", "demo_data/scaling_up.csv",
                      savefile="scaling_up1.2_withsep",
                      dil_coeff=1.2,
                      InputRods="demo_data/scaling_seps.csv")

WatershedSegmentation("dataset", "demo_data/scaling_down.csv",
                      savefile="scaling_down1") # default value of dil_coeff is 1.0

WatershedSegmentation("dataset", "demo_data/scaling_down.csv",
                      savefile="scaling_down0.9",
                      dil_coeff=0.9)
```

### Segmentation using anisotropic pixels

This enables to reproduce Figure 20 of the user manual.

```python
WatershedSegmentation("image", "demo_data/bananas.tiff",
                      savefile="bananas_pixelsize1-1-6",
                      pixelsize=[1,1,6])
```

### Segmentation using different values of the "smoothing" parameter

This enables to reproduce Figures 21 and 22 of the user manual.

```python
# (Fig 21)
WatershedSegmentation("dataset", "demo_data/smoothing.csv", savefile="smoothing0", smooth_coeff=0)
WatershedSegmentation("dataset", "demo_data/smoothing.csv", savefile="smoothing1") # default value of smooth_coeff is 1.0
WatershedSegmentation("dataset", "demo_data/smoothing.csv", savefile="smoothing5", smooth_coeff=5)

# (Fig 22)
WatershedSegmentation("image", "demo_data/bananas.tiff", savefile="bananas_smoothing0", smooth_coeff=0)
WatershedSegmentation("image", "demo_data/bananas.tiff", savefile="bananas_smoothing1") # default value of smooth_coeff is 1.0
WatershedSegmentation("image", "demo_data/bananas.tiff", savefile="bananas_smoothing10", smooth_coeff=10)
```

### Segmentation using realistic parameters

This enables to reproduce the graphical abstract of the user manual.

```python
WatershedSegmentation("image", "demo_data/mouse_adipose_tissue.tif",
                      savefile="mouse_adipose_tissue_seg",
                      pixelsize=[1.84, 1.84, 1], smooth_coeff=10, MinSize=400000)

WatershedSegmentation("dataset", "dataset_spheres.csv",
                      savefile="dataset_spheres_seg",
                      InputRods="dataset_rods.csv",
                      PeriodicBoundaries=True, xmax=15, ymax=15, zmax=15,
                      resolution=3,
                      dil_coeff=1.5,
                      smooth_coeff=0.5,
                      MinSize=10)
```
